```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
area = np.array([600, 850, 1100, 1250, 1500, 1660, 1800, 2100, 2400, 3500]).reshape(-1,1)
price = np.array([3900000, 5500000, 8800000, 11000000, 13500000, 16000000, 19500000, 25000000, 31000000, 58000000])
```

```python
model= LinearRegression()
model.fit(area,price)
pred_price = model.predict(area)
#new_area = input("Enter area: ")
#float_area = float(new_area)
#area_x = np.array([float_area]).reshape(1,-1)
#newpred_price = model.predict(area_x)
#print("Predicted price: ", newpred_price, "Rs.")
```

```python
plt.figure(figsize=(7,5))
plt.scatter(area, price, color='blue', s=80, label='Actual Data')
plt.plot(area, pred_price, color='red', linewidth=3, label='Regression Line')
plt.title("House vs Price: Linear Regression Visualization")
plt.xlabel("Area (Sq Ft)")
plt.ylabel("Price (Rs.)")
plt.grid(True)
plt.legend()
plt.show()
```
